```python
import pandas as pd
from pydataset import data

# Load mtcars dataset
df = data('mtcars')
```

```python
df.head()
```

## 4.1 Add Two Columns
- Create a new column `mpg_hp_sum` = mpg + hp using a lambda function.

```python
df['mpg_hp_sum'] = df.apply(lambda row: row['mpg'] + row['hp'], axis=1)
df[['mpg', 'hp', 'mpg_hp_sum']].head()
```

## 4.2 Conditional Label
- Create a new column `mpg_label`:
  - "High" if mpg > 20
  - "Low" otherwise
- Use `apply()` with a lambda.

```python
df['mpg_label'] = df['mpg'].apply(lambda x: "High" if x > 20 else "Low")
df[['mpg', 'mpg_label']].head()
```

## 4.3 Multiply Column Values
- Double all `hp` values using `map()` with a lambda.

*Note: to avoid overwriting `hp` (which later tasks such as 4.7 and 4.10 rely on), the doubled values are stored in a new column `hp_doubled` rather than replacing `hp` in place.*

```python
df['hp_doubled'] = df['hp'].map(lambda x: x * 2)
df[['hp', 'hp_doubled']].head()
```

## 4.4 Filter Rows
- Get a list of all `mpg` values greater than 25 using `filter()` with a lambda.

```python
mpg_gt_25 = list(filter(lambda x: x > 25, df['mpg']))
print(mpg_gt_25)
```

## 4.5 Reduce to Total
- Use `reduce()` with a lambda to sum all `cyl` values in the dataset.

```python
from functools import reduce

total_cyl = reduce(lambda a, b: a + b, df['cyl'])
print("Total cylinders across all cars:", total_cyl)
```

## 4.6 Sort by Custom Rule
- Sort the dataset by the reverse of the car names' length using `sorted()` with a lambda on the index.

```python
sorted_names = sorted(df.index, key=lambda name: -len(name))
df_sorted_by_name_len = df.loc[sorted_names]
df_sorted_by_name_len.head(10)
```

## 4.7 Multi-Column Condition
- Create a column `efficient_power` = "Yes" if mpg > 20 and hp > 100, else "No".

```python
df['efficient_power'] = df.apply(lambda row: "Yes" if (row['mpg'] > 20 and row['hp'] > 100) else "No", axis=1)
df[['mpg', 'hp', 'efficient_power']].head(10)
```

## 4.8 Normalize a Column
- Normalize `mpg` between 0 and 1 using a lambda inside `apply()`:
- normalized = (x - min) / (max - min)

```python
mpg_min = df['mpg'].min()
mpg_max = df['mpg'].max()

df['mpg_normalized'] = df['mpg'].apply(lambda x: (x - mpg_min) / (mpg_max - mpg_min))
df[['mpg', 'mpg_normalized']].head(10)
```

## 4.9 Extract First Word
- Extract the first word from each car name (index) into a new column `car_make` using `map()` with a lambda.

```python
df['car_make'] = df.index.map(lambda name: name.split()[0])
df[['car_make']].head(10)
```

## 4.10 Square and Add
- Create a column `hp_mpg_calc` = (hp² + mpg²) using `apply()` with a lambda.

```python
df['hp_mpg_calc'] = df.apply(lambda row: row['hp'] ** 2 + row['mpg'] ** 2, axis=1)
df[['hp', 'mpg', 'hp_mpg_calc']].head(10)
```

## 4.11 Filter by Col values
- Filter cars with mpg > 25.

```python
mpg_over_25 = df[df['mpg'] > 25]
mpg_over_25[['mpg', 'hp']]
```

## 4.12 Filter and sorted car names by horsepower
- From above, return a list of car names sorted by horsepower.

```python
car_names_sorted_by_hp = mpg_over_25.sort_values('hp').index.tolist()
print(car_names_sorted_by_hp)
```
