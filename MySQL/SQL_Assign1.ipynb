```python
# SQL - PYTHON Connect
from sqlalchemy import create_engine, text
import pymysql
import pandas as pd
import os
```

```python
# Connect to the piit MySQL database (root user, password Mysql@123 per assignment instructions)
engine1 = create_engine("mysql+pymysql://root:Mysql%40123@localhost/piit")

# function to run the query with parameters or query
def q(sql, **params):
    with engine1.connect() as conn:
        return pd.read_sql(text(sql), conn, params=params)
```

## Write the queries for the following

### 1. List the names of analysts and salesmen.

```python
# MYSQL Query
q("""SELECT ename, job FROM emp WHERE job IN ('ANALYST','SALESMAN')""")
```

```python
# Equivalent pandas code
result = emp[emp['job'].isin(['ANALYST', 'SALESMAN'])][['ename', 'job']]
result
```

### 2. List details of employees who have joined before 30 Sep 81.

```python
# MYSQL Query
q("""SELECT * FROM emp WHERE hiredate < '1981-09-30'""")
```

```python
# Equivalent pandas code
result = emp[emp['hiredate'] < '1981-09-30']
result
```

### 3. List names of employees who are not managers.

```python
# MYSQL Query
q("""SELECT ename, job FROM emp WHERE job <> 'MANAGER'""")
```

```python
# Equivalent pandas code
result = emp[emp['job'] != 'MANAGER'][['ename', 'job']]
result
```

### 4. List the names of employees whose employee numbers are 7369, 7521, 7839, 7934, 7788.

```python
# MYSQL Query
q("""SELECT empno, ename FROM emp WHERE empno IN (7369, 7521, 7839, 7934, 7788)""")
```

```python
# Equivalent pandas code
result = emp[emp['empno'].isin([7369, 7521, 7839, 7934, 7788])][['empno', 'ename']]
result
```

### 5. List employees not belonging to department 30, 40, or 10.

```python
# MYSQL Query
q("""SELECT * FROM emp WHERE deptno NOT IN (30, 40, 10)""")
```

```python
# Equivalent pandas code
result = emp[~emp['deptno'].isin([30, 40, 10])]
result
```

### 6. List employee names for those who have joined between 30 June and 31 Dec '81.

```python
# MYSQL Query
q("""SELECT ename, hiredate FROM emp WHERE hiredate BETWEEN '1981-06-30' AND '1981-12-31'""")
```

```python
# Equivalent pandas code
result = emp[(emp['hiredate'] >= '1981-06-30') & (emp['hiredate'] <= '1981-12-31')][['ename', 'hiredate']]
result
```

### 7. List the different designations in the company.

```python
# MYSQL Query
q("""SELECT DISTINCT job FROM emp""")
```

```python
# Equivalent pandas code
result = pd.DataFrame({'job': emp['job'].unique()})
result
```

### 8. List the names of employees who are not eligible for commission.

```python
# MYSQL Query
q("""SELECT ename FROM emp WHERE comm IS NULL""")
```

```python
# Equivalent pandas code
result = emp[emp['comm'].isna()][['ename']]
result
```

### 9. List the name and designation of the employee who does not report to anybody.

```python
# MYSQL Query
q("""SELECT ename, job FROM emp WHERE mgr IS NULL""")
```

```python
# Equivalent pandas code
result = emp[emp['mgr'].isna()][['ename', 'job']]
result
```

### 10. List the employees not assigned to any department.

```python
# MYSQL Query
q("""SELECT * FROM emp WHERE deptno IS NULL""")
```

```python
# Equivalent pandas code
result = emp[emp['deptno'].isna()]
result
```

### 11. List the employees who are eligible for commission.

```python
# MYSQL Query
q("""SELECT * FROM emp WHERE comm IS NOT NULL""")
```

```python
# Equivalent pandas code
result = emp[emp['comm'].notna()]
result
```

### 12. List employees whose names either start or end with "S".

```python
# MYSQL Query
q("""SELECT ename FROM emp WHERE ename LIKE 'S%' OR ename LIKE '%S'""")
```

```python
# Equivalent pandas code
result = emp[emp['ename'].str.startswith('S') | emp['ename'].str.endswith('S')][['ename']]
result
```

### 13. List names of employees whose names have "i" as the second character.

```python
# MYSQL Query
q("""SELECT ename FROM emp WHERE ename LIKE '_I%'""")
```

```python
# Equivalent pandas code
result = emp[emp['ename'].str[1].str.upper() == 'I'][['ename']]
result
```

### 14. List the number of employees working with the company.

```python
# MYSQL Query
q("""SELECT COUNT(*) AS num_employees FROM emp""")
```

```python
# Equivalent pandas code
result = pd.DataFrame({'num_employees': [len(emp)]})
result
```

### 15. List the number of designations available in the EMP table.

```python
# MYSQL Query
q("""SELECT COUNT(DISTINCT job) AS num_designations FROM emp""")
```

```python
# Equivalent pandas code
result = pd.DataFrame({'num_designations': [emp['job'].nunique()]})
result
```

### 16. List the total salaries paid to the employees.

```python
# MYSQL Query
q("""SELECT SUM(sal) AS total_salary FROM emp""")
```

```python
# Equivalent pandas code
result = pd.DataFrame({'total_salary': [emp['sal'].sum()]})
result
```

### 17. List the maximum, minimum and average salary in the company.

```python
# MYSQL Query
q("""SELECT MAX(sal) AS max_sal, MIN(sal) AS min_sal, AVG(sal) AS avg_sal FROM emp""")
```

```python
# Equivalent pandas code
result = pd.DataFrame({'max_sal': [emp['sal'].max()], 'min_sal': [emp['sal'].min()], 'avg_sal': [emp['sal'].mean()]})
result
```

### 18. List the maximum salary paid to a salesman.

```python
# MYSQL Query
q("""SELECT MAX(sal) AS max_salesman_sal FROM emp WHERE job = 'SALESMAN'""")
```

```python
# Equivalent pandas code
result = pd.DataFrame({'max_salesman_sal': [emp[emp['job'] == 'SALESMAN']['sal'].max()]})
result
```
